### 7. Baseline Model Evaluation (Logistic Regression)
**Goal:** Establish a baseline model performance using normalized main application features.
* **Key Logic:** Stratified train-test split (80/20), median imputation, MinMaxScaler, and L2-regularized Logistic Regression.
* **Takeaway:** Baseline Validation **ROC-AUC Score = 0.6878**.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler,StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score


X=main_df.drop(columns=["TARGET","SK_ID_CURR"])
y=main_df["TARGET"]

X_train,X_val,y_train,y_val=train_test_split(X,y,random_state=42,test_size=0.20,stratify=y)

print(f"TRAINING SHAPE:{X_train.shape}")
print(f"VALIDATION SHAPE:{X_val.shape}")

imputer=SimpleImputer(strategy="median")
X_train_imputed=imputer.fit_transform(X_train)
X_val_imputed=imputer.transform(X_val)

scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train_imputed)
X_val_scaled = scaler.transform(X_val_imputed)

log_reg=LogisticRegression(C=0.0001,max_iter=100)
log_reg.fit(X_train_scaled,y_train)

val_preds=log_reg.predict_proba(X_val_scaled)[:,1]
baseline_score=roc_auc_score(y_val,val_preds)
print(f"\n--- Baseline Model ROC-AUC Score: {baseline_score:.4f} ---")
```

### 8. Credit Bureau Relational Feature Aggregations
**Goal:** Incorporate historical credit activity from external credit bureaus (`bureau.csv`).
* **Key Logic:** Group historical records by `SK_ID_CURR` to compute aggregate metrics (`min`, `max`, `sum`, `mean` for debt and credit amounts). Merge back via left-join.
* **Takeaway:** Validation **ROC-AUC increased from 0.6870 to 0.6917 (+0.0047)**.

```python
bureau=pd.read_csv("C:/Users/joe/Desktop/project-1/data/raw/bureau.csv")
print(bureau.head())
print("----------------------------")
print(bureau.describe())
print("----------------------------")
bureau.columns
print("----------------------------")
bureau.info()
```

```python
bureau_cat=pd.get_dummies(bureau.select_dtypes(include=["object","string"]))
bureau_numeric=bureau.select_dtypes(include=["number"])

bureau_processed=pd.concat([bureau_cat,bureau_numeric],axis=1)

bureau_agg=bureau_processed.groupby("SK_ID_CURR").agg({
    "DAYS_CREDIT":["min","max","sum"],
    "CREDIT_DAY_OVERDUE":["max","mean"],
    "AMT_CREDIT_SUM":["max","mean","sum"],
    "AMT_CREDIT_SUM_DEBT":["max","mean","sum"],
    "AMT_CREDIT_SUM_OVERDUE":["max","sum"]
}).reset_index()


bureau_agg.columns = ['_'.join(col).strip('_') for col in bureau_agg.columns.values]
bureau_agg.rename(columns={'SK_ID_CURR_': 'SK_ID_CURR'}, inplace=True)

print(f"Aggregated bureau shape: {bureau_agg.shape}")

bureau_agg.head()


```

```python
main_df_with_bureau=main_df.merge(bureau_agg,on="SK_ID_CURR",how="left")
print(f"New main_df shape after merging bureau: {main_df.shape}")
```

```python
y = main_df_with_bureau['TARGET']
X = main_df_with_bureau.drop(columns=['TARGET', 'SK_ID_CURR'])

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)


imputer = SimpleImputer(strategy='median')
X_train_imputed = imputer.fit_transform(X_train)
X_val_imputed = imputer.transform(X_val)


scaler = MinMaxScaler()
X_train_scaled = scaler.fit_transform(X_train_imputed)
X_val_scaled = scaler.transform(X_val_imputed)


log_reg = LogisticRegression(C=0.0001, max_iter=100)
log_reg.fit(X_train_scaled, y_train)

val_preds = log_reg.predict_proba(X_val_scaled)[:, 1]



new_score = roc_auc_score(y_val, val_preds)
diff = new_score - 0.6870

print(f"--- Previous Baseline Score: 0.6870 ---")
print(f"--- New Score with Bureau Features: {new_score:.4f} ---")
print(f"--- Difference: {diff:+.4f} ---")
```

### 12. XGBoost Hyperparameter Optimization
**Goal:** Optimize tree depth, learning rate, and sampling rates to maximize validation ROC-AUC.
* **Key Logic:** Execute 3-fold cross-validated `RandomizedSearchCV` across candidate parameters.
* **Takeaway:** Tuned **ROC-AUC improved to 0.7677**, achieving model performance.

```python
from sklearn.model_selection import RandomizedSearchCV
from xgboost import XGBClassifier
from sklearn.metrics import roc_auc_score


param_dist = {
    "n_estimators": [100, 200, 300],
    "max_depth": [3, 4, 5, 6, 7, 8],
    "learning_rate": [0.01, 0.03, 0.05, 0.1],
    'subsample': [0.7, 0.8, 0.9],
    'colsample_bytree': [0.7, 0.8, 0.9],
    'min_child_weight': [1, 3, 5]
}


xgb_base = XGBClassifier(
    eval_metric='auc',
    random_state=42,
    n_jobs=-1
)

xgb_search = RandomizedSearchCV(
    estimator=xgb_base,
    param_distributions=param_dist,
    n_iter=10,
    scoring="roc_auc",
    cv=3,
    verbose=1,
    random_state=42,
    n_jobs=-1
)


xgb_search.fit(X_train_imputed, y_train)


best_xgb = xgb_search.best_estimator_
tuned_preds = best_xgb.predict_proba(X_val_imputed)[:, 1]
tuned_score = roc_auc_score(y_val, tuned_preds)
diff_tuned = tuned_score - 0.7659

print(f"\n--- Best Parameters Found: {xgb_search.best_params_} ---")
print(f"--- Previous XGBoost Score: 0.7659 ---")
print(f"--- Tuned XGBoost ROC-AUC Score: {tuned_score:.4f} ---")
print(f"--- Difference: {diff_tuned:+.4f} ---")
```

### 13. Statistical Threshold Optimization (F1-Score)
**Goal:** Evaluate precision-recall trade-offs and confusion matrix under statistical metrics.
* **Key Logic:** Calculate Precision-Recall curves to find maximum F1-Score threshold.
* **Takeaway:** Statistical optimal threshold located around ~$0.25$.

```python
from sklearn.metrics import precision_recall_curve,f1_score

precisions,recalls,thresholds=precision_recall_curve(y_val,tuned_preds)

f1_scores=2 * (precisions * recalls) / (precisions + recalls + 1e-10)
best_threshold_idx=np.argmax(f1_scores)
best_threshold=thresholds[best_threshold_idx]
best_f1=f1_scores[best_threshold_idx]

print(f"--- Optimal Decision Threshold: {best_threshold:.4f} ---")
print(f"--- Best F1-Score at this Threshold: {best_f1:.4f} ---")
```

```python
from sklearn.metrics import confusion_matrix

y_pred_optimal=(tuned_preds >= best_threshold).astype(int)

cm=confusion_matrix(y_val,y_pred_optimal)

plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Repaid (0)', 'Default (1)'],
            yticklabels=['Repaid (0)', 'Default (1)'])

plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title(f'Confusion Matrix at Optimal Threshold ({best_threshold:.4f})')
plt.tight_layout()
plt.show()
```

```python

gain_tn=500
loss_fn=-10000
loss_fp=-500
gain_tp=0

threshold_range = np.linspace(0.05, 0.50, 100)
net_profits = []

for th in threshold_range:
    preds = (tuned_preds >= th).astype(int)
    
   
    tn = np.sum((y_val == 0) & (preds == 0))
    fp = np.sum((y_val == 0) & (preds == 1))
    fn = np.sum((y_val == 1) & (preds == 0))
    tp = np.sum((y_val == 1) & (preds == 1))
    

    total_profit = (tn * gain_tn) + (fp * loss_fp) + (fn * loss_fn) + (tp * gain_tp)
    net_profits.append(total_profit)


best_biz_idx = np.argmax(net_profits)
best_biz_threshold = threshold_range[best_biz_idx]
max_profit = net_profits[best_biz_idx]

print(f"--- Business-Optimal Threshold: {best_biz_threshold:.4f} ---")
print(f"--- Maximum Net Profit: ${max_profit:,.2f} ---")


plt.figure(figsize=(9, 5))
plt.plot(threshold_range, net_profits, color='green', linewidth=2, label='Net Profit ($)')
plt.axvline(best_biz_threshold, color='red', linestyle='--', label=f'Optimal Th ({best_biz_threshold:.3f})')
plt.xlabel('Decision Threshold')
plt.ylabel('Net Business Profit ($)')
plt.title('Business Profitability vs. Decision Threshold')
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

### 14. Financial Cost-Matrix & Profit Optimization
**Goal:** Align decision-making with bank economics rather than purely statistical metrics.
* **Financial Parameters:**
  * Gain (True Negative - Approved Good Loan): +$1,500
  * Cost (False Positive - Rejected Good Applicant): -$300
  * Loss (False Negative - Approved Defaulter): -$4,000
* **Takeaway:** Optimal Financial Decision Threshold = **0.1773**. Yields a maximum net portfolio profit of **~$64.28M**.

```python

gain_tn = 1500      
loss_fp = -300      
loss_fn = -4000   
gain_tp = 0         
net_profits_real = []

for th in threshold_range:
    preds = (tuned_preds >= th).astype(int)
    
    tn = np.sum((y_val == 0) & (preds == 0))
    fp = np.sum((y_val == 0) & (preds == 1))
    fn = np.sum((y_val == 1) & (preds == 0))
    tp = np.sum((y_val == 1) & (preds == 1))
    
    profit = (tn * gain_tn) + (fp * loss_fp) + (fn * loss_fn) + (tp * gain_tp)
    net_profits_real.append(profit)

best_idx_real = np.argmax(net_profits_real)
best_th_real = threshold_range[best_idx_real]
max_profit_real = net_profits_real[best_idx_real]

print(f"--- Realistic Business Threshold: {best_th_real:.4f} ---")
print(f"--- Realistic Net Profit: ${max_profit_real:,.2f} ---")
```

```python
import matplotlib.pyplot as plt


plt.figure(figsize=(10, 6))


plt.plot(threshold_range, net_profits_real, color='#2ca02c', linewidth=2.5, label='Realistic Net Profit ($)')


plt.axvline(best_th_real, color='#d62728', linestyle='--', linewidth=2, 
            label=f'Optimal Threshold ({best_th_real:.4f})')

plt.scatter(best_th_real, max_profit_real, color='#d62728', s=100, zorder=5)
plt.annotate(f'Max Profit: ${max_profit_real:,.2f}', 
             xy=(best_th_real, max_profit_real), 
             xytext=(best_th_real + 0.03, max_profit_real - 2000000),
             arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=6),
             fontsize=10, fontweight='bold')

plt.xlabel('Decision Threshold', fontsize=12)
plt.ylabel('Net Business Profit ($)', fontsize=12)
plt.title('Realistic Business Profitability vs. Decision Threshold', fontsize=14, pad=15)
plt.grid(True, alpha=0.3)
plt.legend(loc='lower center', fontsize=11)
plt.tight_layout()
plt.show()
```

```python
import joblib
import pandas as pd
import numpy as np

def run_inference(new_data_path):
 
    model = joblib.load('credit_risk_xgb_model.pkl')
    imputer = joblib.load('imputer.pkl')
    feature_names = joblib.load('feature_names.pkl')
    
  
    df = pd.read_csv(new_data_path)
    X = df[feature_names]
    X_imputed = imputer.transform(X)
    
   
    probs = model.predict_proba(X_imputed)[:, 1]
    
    
    BUSINESS_THRESHOLD = 0.1773
    decisions = (probs >= BUSINESS_THRESHOLD).astype(int)
    
    results = pd.DataFrame({
        'SK_ID_CURR': df['SK_ID_CURR'],
        'Default_Probability': probs,
        'Action': np.where(decisions == 1, 'REJECT (High Risk)', 'APPROVE')
    })
    
    return results

if __name__ == '__main__':
    print("--- Running Inference Pipeline ---")
   
```
